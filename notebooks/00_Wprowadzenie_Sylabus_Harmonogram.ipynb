### 2. Zakres merytoryczny (Moduły tematyczne)

* **Moduł I: Inżynieria danych i eksploracyjna analiza danych (EDA)**
  * Pozyskiwanie danych z otwartych API (UN OCHA HDX, kanały RSS WHO / Africa CDC).
  * Architektura medalionowa: izolacja warstwy surowej (*Bronze*), czyszczenie i standaryzacja (*Silver*), agregacja cech (*Gold*).
  * Diagnostyka jakości danych: braki danych (MCAR, MAR, MNAR), imputacja wartości, obsługa niespójności geograficznych i czasowych, zjawisko opóźnienia raportowania w strefach konfliktów.
* **Moduł II: Wykrywanie anomalii i punktów zwrotnych (Anomaly & Changepoint Detection)**
  * Identyfikacja obserwacji odstających w szeregach czasowych i danych przestrzennych.
  * Algorytmy bez nadzoru: *Isolation Forest*, *Local Outlier Factor (LOF)*, grupowanie gęstościowe *DBSCAN*.
  * Wykrywanie nagłych skoków zachorowań i wczesne ostrzeganie przed utratą kontroli nad ogniskiem epidemicznym.
* **Moduł III: Eksploracja grafów i dynamika sieci (Graph Mining & Network Analysis)**
  * Reprezentacja szlaków migracyjnych, powiązań administracyjnych i sieci kontaktów w postaci grafów ważonych.
  * Miary centralności węzłów (*Degree*, *Betweenness*, *Closeness*, *PageRank*) w identyfikacji kluczowych węzłów transmisji.
  * Analiza odporności sieci i optymalizacja tras logistycznych (zimny łańcuch dostaw) w warunkach dynamicznych blokad drogowych.
* **Moduł IV: Modelowanie predykcyjne i klasyfikacja w warunkach niezrównoważenia klas**
  * Drzewa decyzyjne, lasy losowe (*Random Forest*), wzmacnianie gradientowe (*XGBoost / LightGBM*).
  * Problem silnie niezrównoważonych zbiorów danych (rzadkie wystąpienia ognisk) — dlaczego zwykła dokładność (*Accuracy*) zawodzi.
  * Dobór właściwych metryk ewaluacji: krzywa *Precision-Recall (PR-AUC)*, *ROC-AUC*, czułość (*Recall*), miara $F_\beta$, walidacja krzyżowa z zachowaniem struktury czasowo-przestrzennej.
* **Moduł V: Eksploracja tekstu i wywiad jawnoźródłowy (Text Mining & Medical OSINT)**
  * Automatyczne przetwarzanie nieustrukturyzowanych komunikatów terenowych i raportów WHO.
  * Wektoryzacja tekstu (TF-IDF, osadzenia wektorowe), wydobywanie encji nazwanych (NER — nazwy miejscowości, liczby przypadków, incydenty bezpieczeństwa).
